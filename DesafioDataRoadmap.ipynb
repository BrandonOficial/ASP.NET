```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração global de UI para gráficos mais limpos e elegantes
plt.style.use('ggplot')
sns.set_theme(style="whitegrid")

# Instanciando o DataFrame com nomenclatura clara
tips_data = sns.load_dataset("tips")
display(tips_data.head())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=tips_data, x='total_bill', bins=20, color='skyblue', kde=True)
plt.title('Distribuição do Valor Total da Conta', fontsize=14)
plt.xlabel('Valor da Conta ($)')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=tips_data, x='day', y='tip', palette='pastel')
plt.title('Média de Gorjeta por Dia', fontsize=14)
plt.xlabel('Dia da Semana')
plt.ylabel('Média da Gorjeta ($)')
plt.show()
```

```python
# Filtragem rigorosa de tipos para evitar quebra de contrato na matriz de correlação
numeric_columns = tips_data.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_columns.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Heatmap de Correlações', fontsize=14)

# Persistindo o artefato visual no ambiente virtual
plt.savefig('heatmap_correlacao.png', dpi=300, bbox_inches='tight')
plt.show()
```
